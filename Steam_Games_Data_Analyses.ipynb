## Step 4: Data Cleaning

We clean on a **copy** of the data (`df`) so the original (`raw`) stays safe for comparison.

```python
df = raw.copy()

# make all column names small letters and remove spaces
df.columns = df.columns.str.strip().str.lower()

# check that the columns we need are present
needed = ["appid", "name", "release_date", "platforms", "genres",
          "positive_ratings", "negative_ratings", "average_playtime", "price"]
for c in needed:
    if c not in df.columns:
        print("WARNING: column not found ->", c)

print("Columns:", list(df.columns))
```

### 4.1 Missing Values

**Check missing values in every column (before cleaning)**

```python
missing = df.isnull().sum()
percent = (df.isnull().sum() / len(df) * 100).round(2)
pd.DataFrame({"missing_values": missing, "percent": percent})
```

**Fill the missing values**
- Text columns (developer, publisher, genres, etc.) are filled with `"Unknown"`
- Number columns are filled with the **median** (the median is not affected by very big or very small values)

```python
text_columns = ["name", "developer", "publisher", "platforms", "categories", "genres", "steamspy_tags"]

for col in text_columns:
    if col in df.columns:
        df[col] = df[col].fillna("Unknown")
        df[col] = df[col].astype(str).str.strip()
        df[col] = df[col].replace("", "Unknown")

for col in df.select_dtypes(include="number").columns:
    df[col] = df[col].fillna(df[col].median())

print("Missing values filled")
```

**Check again (after cleaning)**

```python
print("Total missing values now:", df.isnull().sum().sum())
df.isnull().sum()
```

### 4.2 Duplicates

```python
print("Rows before removing duplicates:", df.shape[0])
print("Fully duplicate rows           :", df.duplicated().sum())
print("Repeated appid (game id)       :", df.duplicated(subset="appid").sum())
print("Repeated game names            :", df.duplicated(subset="name").sum(), "(only checked, not removed)")
```

```python
df = df.drop_duplicates()
df = df.drop_duplicates(subset="appid", keep="first")
df = df.reset_index(drop=True)

print("Rows after removing duplicates :", df.shape[0])
print("Duplicate rows left            :", df.duplicated().sum())
```

### 4.3 Data Types

**Data types before changing**

```python
df.dtypes
```

**Change the data types**
- Number columns are converted with `pd.to_numeric`
- `release_date` is converted to a date, and `release_year` is made from it
- `owners` looks like `20000-50000`, so we split it into low, high and middle values

```python
number_columns = ["appid", "required_age", "achievements", "positive_ratings",
                  "negative_ratings", "average_playtime", "median_playtime", "price"]

for col in number_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")
        df[col] = df[col].fillna(df[col].median())

# date column
df["release_date"] = pd.to_datetime(df["release_date"], errors="coerce")
df["release_year"] = df["release_date"].dt.year
print("Dates that could not be read:", df["release_date"].isnull().sum())

# owners column (example: 20000-50000)
if "owners" in df.columns:
    owners_split = df["owners"].astype(str).str.replace(",", "").str.extract(r"(\d+)\s*-\s*(\d+)")
    df["owners_low"] = owners_split[0].astype(float)
    df["owners_high"] = owners_split[1].astype(float)
    df["owners_mid"] = (df["owners_low"] + df["owners_high"]) / 2
```

**New columns that we need for the analysis**

```python
df["total_ratings"] = df["positive_ratings"] + df["negative_ratings"]
df["positive_pct"] = df["positive_ratings"] / df["total_ratings"] * 100     # NaN when a game has 0 ratings

# platform columns (True / False)
df["windows"] = df["platforms"].str.contains("windows", case=False)
df["mac"] = df["platforms"].str.contains("mac", case=False)
df["linux"] = df["platforms"].str.contains("linux", case=False)

print("New columns added: release_year, owners_low/high/mid, total_ratings, positive_pct, windows, mac, linux")
```

**Remove impossible values (negative numbers)**

```python
print("Negative prices         :", (df["price"] < 0).sum())
print("Negative positive ratings:", (df["positive_ratings"] < 0).sum())
print("Negative playtime       :", (df["average_playtime"] < 0).sum())

df = df[(df["price"] >= 0) & (df["positive_ratings"] >= 0) &
        (df["negative_ratings"] >= 0) & (df["average_playtime"] >= 0)]
df = df.reset_index(drop=True)
print("Rows now:", df.shape[0])
```

**Data types after changing**

### 4.4 Outliers

We use the **IQR method**:
- IQR = Q3 - Q1
- Upper limit = Q3 + 1.5 x IQR
- A value above the upper limit is an outlier

A price of 0 means a free game and a playtime of 0 means no playtime was recorded, so we leave the zeros out when we calculate Q1 and Q3.

**Boxplots before handling outliers (log scale because some values are very big)**

```python
plt.figure(figsize=(13, 4))

plt.subplot(1, 3, 1)
plt.boxplot(df[df["price"] > 0]["price"])
plt.yscale("log")
plt.title("Price")

plt.subplot(1, 3, 2)
plt.boxplot(df[df["average_playtime"] > 0]["average_playtime"])
plt.yscale("log")
plt.title("Average playtime")

plt.subplot(1, 3, 3)
plt.boxplot(df[df["positive_ratings"] > 0]["positive_ratings"])
plt.yscale("log")
plt.title("Positive ratings")

plt.tight_layout()
plt.show()
```

**Outliers in price**

```python
price_values = df[df["price"] > 0]["price"]
Q1 = price_values.quantile(0.25)
Q3 = price_values.quantile(0.75)
IQR = Q3 - Q1
price_upper = Q3 + 1.5 * IQR

df["price_outlier"] = df["price"] > price_upper

print("Q1 =", Q1, "| Q3 =", Q3, "| IQR =", IQR)
print("Upper limit for price:", round(price_upper, 2))
print("Price outliers:", df["price_outlier"].sum())
```

**Outliers in average playtime**

```python
play_values = df[df["average_playtime"] > 0]["average_playtime"]
Q1 = play_values.quantile(0.25)
Q3 = play_values.quantile(0.75)
IQR = Q3 - Q1
play_upper = Q3 + 1.5 * IQR

df["playtime_outlier"] = df["average_playtime"] > play_upper

print("Q1 =", Q1, "| Q3 =", Q3, "| IQR =", IQR)
print("Upper limit for playtime (minutes):", round(play_upper, 2))
print("Playtime outliers:", df["playtime_outlier"].sum())
```

**Outliers in positive ratings**

```python
rating_values = df[df["positive_ratings"] > 0]["positive_ratings"]
Q1 = rating_values.quantile(0.25)
Q3 = rating_values.quantile(0.75)
IQR = Q3 - Q1
rating_upper = Q3 + 1.5 * IQR

df["ratings_outlier"] = df["positive_ratings"] > rating_upper

print("Q1 =", Q1, "| Q3 =", Q3, "| IQR =", IQR)
print("Upper limit for positive ratings:", round(rating_upper, 2))
print("Ratings outliers:", df["ratings_outlier"].sum())
```

**How we handle the outliers**

The outliers are only **marked** in new columns (`price_outlier`, `playtime_outlier`, `ratings_outlier`). We do **not** delete them because they are real games (for example very expensive games or games played for hundreds of hours). They stay in the top-10 charts and the scatter plot. The histograms and the box plot leave them out so that the main pattern is easy to see.

```python
outlier_table = pd.DataFrame({
    "outliers": [df["price_outlier"].sum(), df["playtime_outlier"].sum(), df["ratings_outlier"].sum()],
    "percent": [df["price_outlier"].mean() * 100, df["playtime_outlier"].mean() * 100, df["ratings_outlier"].mean() * 100]
}, index=["price", "average_playtime", "positive_ratings"])
outlier_table.round(2)
```

## Step 5: Graphs

Eight graphs are made as planned in the project proposal. Each graph is saved in the `charts` folder.

### Graph 1: Top 10 Most Common Game Genres (Horizontal Bar Chart)

```python
# a game can have many genres like "Action;Indie", so we split them and count each one
genre_df = df[["appid", "genres", "price", "price_outlier"]].copy()
genre_df["genre"] = genre_df["genres"].str.split(";")
genre_df = genre_df.explode("genre")
genre_df["genre"] = genre_df["genre"].str.strip()
genre_df = genre_df[genre_df["genre"] != "Unknown"]

genre_counts = genre_df["genre"].value_counts()
print(genre_counts.head(10))
```

```python
top10_genres = genre_counts.head(10).sort_values()

plt.figure(figsize=(9, 5))
plt.barh(top10_genres.index, top10_genres.values, color="steelblue")
for i in range(len(top10_genres)):
    plt.text(top10_genres.values[i], i, " " + str(top10_genres.values[i]), va="center")
plt.title("Top 10 Most Common Game Genres on Steam")
plt.xlabel("Number of games")
plt.ylabel("Genre")
plt.tight_layout()
plt.savefig("charts/graph1_top_genres.png", dpi=150)
plt.show()

print("Most common genre:", genre_counts.index[0], "with", genre_counts.iloc[0], "games")
```

### Graph 2: Distribution of Game Prices (Histogram)

```python
paid_games = df[df["price"] > 0]
prices = paid_games[paid_games["price_outlier"] == False]["price"]

plt.figure(figsize=(9, 5))
plt.hist(prices, bins=np.arange(0, prices.max() + 1, 1), color="royalblue", edgecolor="white")   # each bar = 1 price unit
plt.axvline(prices.median(), color="red", linestyle="--", label="Median = " + str(round(prices.median(), 2)))
plt.title("Distribution of Game Prices (paid games, outliers removed)")
plt.xlabel("Price")
plt.ylabel("Number of games")
plt.legend()
plt.tight_layout()
plt.savefig("charts/graph2_price_distribution.png", dpi=150)
plt.show()

print("Free games (not in graph)          :", (df["price"] == 0).sum())
print("Price outliers (not in graph)      :", paid_games["price_outlier"].sum())
print("Median price of paid games         :", round(paid_games["price"].median(), 2))
```

### Graph 3: Top 10 Games by Positive Ratings (Horizontal Bar Chart)

```python
top10_rated = df.nlargest(10, "positive_ratings")[["name", "positive_ratings"]]
print(top10_rated)
```

```python
top10_rated = top10_rated.sort_values("positive_ratings")

plt.figure(figsize=(10, 5))
plt.barh(top10_rated["name"], top10_rated["positive_ratings"], color="seagreen")
plt.title("Top 10 Games by Positive Ratings")
plt.xlabel("Positive ratings")
plt.ylabel("Game")
plt.tight_layout()
plt.savefig("charts/graph3_top_positive_ratings.png", dpi=150)
plt.show()
```

### Graph 4: Platform Support Distribution (Pie Chart)

```python
def platform_type(row):
    if row["windows"] and row["mac"] and row["linux"]:
        return "Windows + Mac + Linux"
    elif row["windows"] and row["mac"]:
        return "Windows + Mac"
    elif row["windows"] and row["linux"]:
        return "Windows + Linux"
    elif row["windows"]:
        return "Windows only"
    else:
        return "Other"

df["platform_type"] = df.apply(platform_type, axis=1)
platform_counts = df["platform_type"].value_counts()
print(platform_counts)
```

```python
plt.figure(figsize=(7, 6))
plt.pie(platform_counts.values, autopct=lambda p: str(round(p, 1)) + "%" if p > 3 else "", startangle=90)   # small slices have no label
plt.legend(platform_counts.index, title="Platform support", loc="center left", bbox_to_anchor=(1, 0.5))
plt.title("Platform Support Distribution")
plt.tight_layout()
plt.savefig("charts/graph4_platform_support.png", dpi=150, bbox_inches="tight")
plt.show()

print("Games that support Windows:", round(df["windows"].mean() * 100, 1), "%")
```

### Graph 5: Game Price vs. Positive Ratings (Scatter Plot)

```python
rated = df[df["positive_ratings"] > 0]

plt.figure(figsize=(9, 5.5))
plt.scatter(rated["price"], rated["positive_ratings"], s=8, alpha=0.3, color="purple")
plt.yscale("log")                         # log scale because ratings are very different in size
plt.xlim(-0.5, rated["price"].quantile(0.995))
plt.title("Game Price vs. Positive Ratings")
plt.xlabel("Price")
plt.ylabel("Positive ratings (log scale)")
plt.tight_layout()
plt.savefig("charts/graph5_price_vs_ratings.png", dpi=150)
plt.show()

correlation = rated["price"].corr(rated["positive_ratings"], method="spearman")
print("Correlation between price and positive ratings (Spearman):", round(correlation, 2))
```

### Graph 6: Number of Games Released per Year (Line Chart)

```python
games_per_year = df.dropna(subset=["release_year"]).groupby("release_year").size()
games_per_year.index = games_per_year.index.astype(int)
print(games_per_year)
```

```python
plt.figure(figsize=(10, 5))
plt.plot(games_per_year.index, games_per_year.values, marker="o", color="darkorange")
plt.title("Number of Games Released on Steam per Year")
plt.xlabel("Release year")
plt.ylabel("Number of games")
plt.grid(True)
last_date = df["release_date"].max()
note = "Note: data ends in " + last_date.strftime("%B %Y") + ", so " + str(last_date.year) + " is not a full year"
plt.text(games_per_year.index[0], games_per_year.max() * 0.95, note, fontsize=9)
plt.tight_layout()
plt.savefig("charts/graph6_releases_per_year.png", dpi=150)
plt.show()

print("Year with most releases:", games_per_year.idxmax(), "(", games_per_year.max(), "games )")
print("Latest release date in the data:", last_date.date(), "(the last year may be incomplete)")
```

### Graph 7: Distribution of Average Playtime (Histogram)

```python
played = df[df["average_playtime"] > 0]
hours = played[played["playtime_outlier"] == False]["average_playtime"] / 60     # minutes to hours

plt.figure(figsize=(9, 5))
plt.hist(hours, bins=40, color="green", edgecolor="white")
plt.axvline(hours.median(), color="red", linestyle="--", label="Median = " + str(round(hours.median(), 1)) + " hours")
plt.title("Distribution of Average Playtime (outliers removed)")
plt.xlabel("Average playtime (hours)")
plt.ylabel("Number of games")
plt.legend()
plt.tight_layout()
plt.savefig("charts/graph7_playtime_distribution.png", dpi=150)
plt.show()

print("Games with 0 playtime (not in graph):", (df["average_playtime"] == 0).sum())
print("Playtime outliers (not in graph)    :", played["playtime_outlier"].sum())
```

### Graph 8: Game Prices by Genre (Box Plot)

```python
top5_genres = genre_counts.head(5).index.tolist()
print("Top 5 genres:", top5_genres)

box_data = genre_df[(genre_df["price"] > 0) & (genre_df["price_outlier"] == False) & (genre_df["genre"].isin(top5_genres))]
```

```python
price_lists = []
for g in top5_genres:
    price_lists.append(box_data[box_data["genre"] == g]["price"])

plt.figure(figsize=(9, 5.5))
plt.boxplot(price_lists, patch_artist=True)
plt.xticks(range(1, 6), top5_genres)
plt.title("Game Prices by Genre (top 5 genres, paid games, outliers removed)")
plt.xlabel("Genre")
plt.ylabel("Price")
plt.tight_layout()
plt.savefig("charts/graph8_price_by_genre.png", dpi=150)
plt.show()

box_data.groupby("genre")["price"].describe().loc[top5_genres].round(2)
```

## Step 6: Short Insights Report
